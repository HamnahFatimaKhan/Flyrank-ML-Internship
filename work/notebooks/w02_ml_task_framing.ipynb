## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

* **Limitations of fixed rules:** Simple rules (e.g., `days_since_last_update > 180`) create high rates of false positives because many older pages remain top performers. Fixed thresholds cannot account for complex feature interactions between ranking position, impression drop rate, and page category simultaneously.
* **Why ML wins:** Machine learning decision trees and gradient boosted models capture non-linear trade-offs across multiple features (e.g., a page with high exposure that is slightly stale vs. a low exposure page that is very stale), producing continuous risk scores that accurately rank pages by true urgency.
* **Action Supported:** Content strategists export the top K flagged list into their CMS sprint backlog to schedule editorial updates.
