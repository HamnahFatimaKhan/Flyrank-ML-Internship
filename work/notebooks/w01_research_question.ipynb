## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search Question: Which published content pages are experiencing statistically significant performance decline (impressions/clicks) versus natural variance, and require urgent content updates?

Unit of Analysis: A single page (page_path / content URL) over a 90-day window.

Output: A predicted decline risk score and classification (is_declining_label).

Decision & Action: The content team reviews the high-risk flagged list and allocates editorial budget to rewrite or refresh those specific pages.

Cost of a Wrong Call:

False Positive (flagged declining, but actually fine): Wasted editorial hours and budget updating content that didn't need intervention.

False Negative (missed a declining page): Loss of high-intent organic traffic and conversion potential over months, leading to revenue drop before anyone notices.

Why Data/ML helps: Hand-written rules struggle with distinguishing seasonal noise from systemic rank decay across tens of thousands of pages. Machine learning identifies complex interactions between page age, position decay, and impression drops that simple static thresholds miss.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import subprocess
import pandas as pd
import numpy as np

# Clone repo if missing and navigate to root
REPO_NAME = "Flyrank-ML-Internship"
REPO_URL = f"https://github.com/HamnahFatimaKhan/{REPO_NAME}.git"

if not os.path.exists("data/raw/content_refresh_anonymized.csv"):
    if os.path.exists(f"../../data/raw/content_refresh_anonymized.csv"):
        os.chdir("../..")
    elif not os.path.exists(REPO_NAME):
        subprocess.run(["git", "clone", REPO_URL], check=True)
        os.chdir(REPO_NAME)
    else:
        os.chdir(REPO_NAME)

# Load starter data
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create target label
df["is_declining_label"] = (df["trend_direction"].str.lower() == "down").astype(int)

# Number 1: Overall declining baseline
declining_rate = df["is_declining_label"].mean()
print(f"Real Number 1 - Baseline Declining Rate: {declining_rate:.1%}")

# Number 2: Impact of stale pages (days_since_last_update >= 180)
stale_mask = df["days_since_last_update"] >= 180
stale_decline_rate = df[stale_mask]["is_declining_label"].mean()
print(f"Real Number 2 - Decline Rate for Stale Pages (>=180 days): {stale_decline_rate:.1%}")

# Number 3: Exposure vs Decay (Visible pages with high impressions)
visible_high_imp = df[df["impressions_90d"] >= 500]
high_imp_decline = visible_high_imp["is_declining_label"].mean()
print(f"Real Number 3 - Decline Rate for High Impression Pages (>=500): {high_imp_decline:.1%}")
```
